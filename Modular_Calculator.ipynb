```python
def add(a, b):
    # Returns the sum of two numbers.
    return a + b

def subtract(a, b):
    # Returns the difference of two numbers.
    return a - b

def multiply(a, b):
    # Returns the product of two numbers.
    return a * b

def divide(a,b):
  # Returns the quotient of two numbers.
  return a / b

def exponentiate(a, b):
  # Returns the exponent of two numbers.
  return a ** b

def root(a, b):
  # Returns the base of the power.
  return a ** (1/b)

def log(a, b):
  # Returns the exponent of a power.
  if a <= 0 or b <= 1:
    raise ValueError("Invalid inputs: 'a' must be > 0 and 'b' must be > 1.")

  exponent = 0
  current_value = 1

  while current_value * b <= a:
    current_value *= b
    exponent += 1

  return exponent


def add_call():
  # Activates when addition is called.
  add_num_1 = int(input("Enter the first addend for addition. "))
  add_num_2 = int(input("Enter the second addend for addition. "))
  sum = add(add_num_1, add_num_2)
  print(sum)

def subtract_call():
  # Activates when subtraction is called.
  sub_num_1 = int(input("Enter the subtrahend for subtraction. "))
  sub_num_2 = int(input("Enter the minuend for subtraction. "))
  difference = subtract(sub_num_1, sub_num_2)
  print(difference)

def multiply_call():
  # Activates when multiplication is called.
  mul_num_1 = int(input("Enter the first factor for multiplication. "))
  mul_num_2 = int(input("Enter the second factor for multiplication. "))
  product = multiply(mul_num_1, mul_num_2)
  print(product)

def divide_call():
  # Activates when division is called.
  div_num_1 = int(input("Enter the dividend for division. "))
  div_num_2 = int(input("Enter the divisor for division. "))
  if div_num_2 == 0 :
    print("Division by zero error.")
  else:
    quotient = divide(div_num_1, div_num_2)
  print(quotient)

def exponentiation_call():
  # Activates when exponentiation is called.
  exp_num_1 = int(input("Enter the base for exponentiation. "))
  exp_num_2 = int(input("Enter the exponent for exponentiation. "))
  power = exponentiate(exp_num_1, exp_num_2)
  print(power)

def root_call():
  # Activates when exponential root is called.
  root_num_1 = int(input("Enter the power for exponential root. "))
  root_num_2 = int(input("Enter the exponent for exponential root. "))
  exponent = root(root_num_1, root_num_2)
  print

def log_call():
  # Activates when log is called.
  log_num_1 = int(input("Enter the power for log. "))
  log_num_2 = int(input("Enter the base for log. "))
  base = log(log_num_1, log_num_2)
  print(base)

while True:
    operation = str(input("What operation would you like to perform? (type 'exit' to quit): ")).strip().lower()

    if operation == "exit":
        print("Goodbye!")
        break  # Exits the loop and stops the program

    elif operation == "add":
        add_call()
    elif operation == "subtract":
        subtract_call()
    elif operation == "multiply":
        multiply_call()
    elif operation == "divide":
        divide_call()
    elif operation == "expontiate":
        exponentiation_call()
    elif operation == "root":
        root_call()
    elif operation == "log":
        log_call()
    else:
        print("Invalid input. Please type an operation of mathematics?")
```
